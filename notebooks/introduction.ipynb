# PULSE GUARD

Mental Health Risk Detection for UK School Safeguarding

## Problem

Welfare concerns recorded by different staff members can be difficult to connect over time. Designated Safeguarding Leads (DSLs) need a clear view of these records to identify students who may need support.

## Project Aim

Develop an NLP system that analyses welfare records, flags potential mental health risks, and explains why a record needs attention. Daily summaries will help DSLs prioritise review and connect concerns about the same student.

The system is intended to run on the school's own network, keeping student data on site and supporting staff judgement.

## Project Phases

1. Data collection
2. Exploratory data analysis (EDA)
3. Preprocessing and tokenisation
4. Model training
5. Evaluation
6. Inference
7. Deployment
